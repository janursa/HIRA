```python
import anndata as ad
import pandas as pd
import scanpy as sc

adata = ad.read_h5ad('/vol/projects/jnourisa/datasets/data7_bulk.h5ad') 

sc.pp.neighbors(adata)  # Compute nearest neighbors
sc.tl.umap(adata)  # Compute UMAP

sc.pl.umap(adata, color="cell_type")  # Change "cell_type" to any metadata column of interest
sc.pl.pca(adata, color="cell_type")  # Change "cell_type" to any metadata column of interest
```

```python
sc.pl.pca(adata, color="cell_count")  # Change "cell_type" to any metadata column of interest
```
